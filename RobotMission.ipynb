```python
import pandas as pd
data=pd.read_csv("Data/robot_mission.csv")

numbers=data["number"].tolist()
rounds=data["round"].tolist()
print(numbers)
print(rounds)
total=0
positions=[]
for i in range(len(numbers)):
  if numbers[i]%2==0:
    total+=numbers[i]
  else:
    total-=numbers[i]
  positions.append(total)
  print("for round No. ("+str(i+1)+") the position is ("+str(positions[i])+")")
```

```python
 import matplotlib.pyplot as plt
 plt.plot(rounds,positions)
 plt.xlabel("Rounds")
 plt.ylabel("Positions")
 plt.title("Robot Mission")
 plt.grid()
 plt.show()
 print("The Total is : " + str(total))
```
